## Parte 0 — O notebook e o ambiente

Um notebook é uma sequência de **células**. Cada uma é de texto (como esta) ou de código.

- **`Ctrl+Enter`** executa e fica onde está.

A coisa mais importante para entender hoje: **a ordem que vale é a de execução, não a da

## Parte 1 — Do C/Java para o Python

### As cinco diferenças que mais pegam

| | C / Java | Python |
|---|---|---|
| Blocos | `{ }` | **indentação** — 4 espaços |
| Tipos | declarados: `int x = 5;` | inferidos: `x = 5` |
| Fim de linha | `;` | nada |
| Booleanos | `true`, `false` | `True`, `False` |
| Nulo | `null` | `None` |

Em Java você escreveria:

```java
int maior(int a, int b) {
    if (a > b) {
        return a;
    }
    return b;
}
```

Em Python:


````python
def maior(a, b):
    if a > b:
        return a
    return b

print(maior(3, 7))
print(maior("banana", "abacate"))   # funciona com qualquer coisa comparável
````

Sem chaves, sem ponto e vírgula, sem tipos. Os **dois pontos** abrem o bloco; a
**indentação** o delimita.

E repare na segunda chamada: a mesma função comparou textos. Em Python o tipo é do *valor*,
não da *variável*.

> ⚠️ **Nunca misture espaço e tabulação.** Configure o editor para inserir 4 espaços ao

### if / elif / else

Note o `elif` — é `else if` abreviado — e a ausência de parênteses na condição.


````python
def classifica(nota):
    if nota >= 9:
        return "excelente"
    elif nota >= 7:
        return "bom"
    elif nota >= 6:
        return "suficiente"
    else:
        return "insuficiente"

for n in [10, 8, 6.5, 4]:
    print(n, "->", classifica(n))
````

Duas coisas que Java não tem:


````python
i = 5

# 1. comparações encadeadas
if 0 <= i < 10:
    print("i está entre 0 e 9")

# 2. operadores lógicos são PALAVRAS
if i > 0 and not (i % 2 == 0):
    print("i é positivo e ímpar")
````

### for percorre ELEMENTOS, não índices

Este é o ponto em que quem vem de C mais tropeça. Em C você escreveria
`for (int i = 0; i < n; i++)`. Em Python, o `for` percorre diretamente os elementos:


````python
cores = ["azul", "verde", "vermelho"]

for cor in cores:            # não precisa de índice
    print(cor)

print("---")
for i in range(3):           # 0, 1, 2
    print(i)

print("---")
for i in range(2, 11, 3):    # início, fim (exclusivo), passo
    print(i)
````

Precisa do índice **e** do valor? Use `enumerate` — não crie um contador na mão:


````python
for i, cor in enumerate(cores):
    print(i, cor)

print("---")
# duas listas em paralelo: zip
notas = [8.5, 7.0, 9.5]
alunos = ["Ana", "Bruno", "Carla"]
for aluno, nota in zip(alunos, notas):
    print(f"{aluno:<8} {nota}")
````

### while

Igual ao que você já conhece. Existe `break` e `continue`.


````python
n, passos = 27, 0
while n != 1:                      # sequência de Collatz
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    passos += 1
print("chegou a 1 em", passos, "passos")
````

Repare no `n // 2 if n % 2 == 0 else 3 * n + 1`: é o operador ternário do Python.
Em Java seria `n % 2 == 0 ? n/2 : 3*n+1`.

## Parte 3 — As quatro coleções

**Esta é a parte mais importante do laboratório.** Escolher a coleção errada é a diferença
entre um programa que roda e um que trava.

| | Sintaxe | Muda depois? | Ordem | Busca `in` | Use quando |
|---|---|---|---|---|---|
| **list** | `[1, 2, 3]` | sim | sim | lenta `O(n)` | sequência que cresce |
| **tuple** | `(1, 2)` | **não** | sim | lenta | registro fixo; **chave de dict** |
| **dict** | `{"a": 1}` | sim | inserção | **rápida `O(1)`** | associar chave a valor |
| **set** | `{1, 2}` | sim | **não** | **rápida `O(1)`** | pertence ou não pertence |

### list


````python
lista = [3, 1, 2]

lista.append(4)          # acrescenta no fim
print(lista)

print("removido:", lista.pop())   # remove e devolve o último
print("tamanho:", len(lista))
print("ordenada (cópia):", sorted(lista))

lista.sort()             # ordena no lugar
print("ordenada no lugar:", lista)
````

**Fatiamento** — funciona em list, tuple e str:


````python
L = [0, 10, 20, 30, 40]

print("L[1:3] =", L[1:3])    # do índice 1 até ANTES do 3
print("L[:2]  =", L[:2])
print("L[2:]  =", L[2:])
print("L[-1]  =", L[-1])     # último
print("L[::-1]=", L[::-1])   # invertido (cópia)
````

### dict — associação chave → valor


````python
idade = {"Ana": 20, "Bruno": 22}

idade["Carla"] = 19          # insere
idade["Ana"] = 21            # atualiza
print(idade)

print("Ana tem", idade["Ana"])
print("Zeca tem", idade.get("Zeca", "idade desconhecida"))   # não dá erro

print()
for nome, anos in idade.items():
    print(f"{nome:<8} {anos}")
````

### set — pertence ou não pertence


````python
visitados = set()

visitados.add((0, 0))
visitados.add((1, 0))
visitados.add((0, 0))        # já estava: não duplica

print(visitados, "->", len(visitados), "elementos")
print("(1,0) já foi visitado?", (1, 0) in visitados)
print("(5,5) já foi visitado?", (5, 5) in visitados)
````

### Por que isso importa — e muito

Duas demonstrações. Primeiro, **por que a tupla pode ser chave e a lista não**:


````python
# tupla como elemento de um set: funciona
ok = {(3, 5), (1, 2)}
print("set de tuplas:", ok)

# lista como elemento de um set: erro
try:
    ruim = {[3, 5]}
except TypeError as e:
    print("erro:", e)
````

A mensagem é `unhashable type: 'list'`. Coisas que **mudam** não podem ser chave, porque a
posição delas na tabela dependeria de um valor que pode mudar depois.

**É por isso que, nos próximos laboratórios, um estado é sempre uma tupla.**

Agora a segunda demonstração: **velocidade**.


````python
import time

N = 20_000
como_lista = list(range(N))
como_set   = set(range(N))

t = time.perf_counter()
for x in range(N):
    x in como_lista
tempo_lista = time.perf_counter() - t

t = time.perf_counter()
for x in range(N):
    x in como_set
tempo_set = time.perf_counter() - t

print(f"buscar {N} vezes em uma list: {tempo_lista:7.3f} s")
print(f"buscar {N} vezes em um  set : {tempo_set:7.3f} s")
print(f"o set foi ~{tempo_lista/tempo_set:.0f}x mais rápido")
````

Guarde esse número. Nos laboratórios de busca, o conjunto `visitados` é consultado uma vez
por vizinho de cada nó expandido — dezenas de milhares de vezes. Com uma lista, o programa
trava; com um set, roda instantaneamente.

**Escolher a estrutura certa não é estilo. É a diferença entre funcionar e não funcionar.**

### Devolver várias coisas

Python não tem "retorno múltiplo" — ele devolve **uma tupla**, e você a desempacota:


````python
def divide(a, b):
    return a // b, a % b        # na verdade devolve a tupla (quociente, resto)

q, r = divide(17, 5)
print(f"17 = 5 * {q} + {r}")
````

### ⚠️ A armadilha do argumento mutável

Esta pega gente experiente. Leia o resultado com atenção:


````python
def errado(item, lista=[]):     # NUNCA faça isso
    lista.append(item)
    return lista

print(errado(1))
print(errado(2))     # esperava [2]?
print(errado(3))
````

O valor padrão é criado **uma única vez**, quando a função é definida — e não a cada
chamada. Todas as chamadas compartilham a mesma lista.

O jeito certo:


````python
def certo(item, lista=None):
    if lista is None:
        lista = []
    lista.append(item)
    return lista

print(certo(1))
print(certo(2))
print(certo(3))
````

## Parte 6 — Armadilhas

### Atribuição não copia


````python
a = [1, 2, 3]
b = a                # NÃO copiou: b e a são o MESMO objeto
b.append(4)
print("a =", a)      # a mudou também

c = a.copy()         # agora sim
c.append(99)
print("a =", a)
print("c =", c)
````

Vale para list, dict e set. Para estruturas aninhadas existe `copy.deepcopy`.

### `==` compara valor; `is` compara identidade


````python
p = [1, 2]
q = [1, 2]

print("p == q ?", p == q)    # mesmo conteúdo
print("p is q ?", p is q)    # objetos diferentes

# com None, use SEMPRE is
valor = None
print("valor is None ?", valor is None)
````
